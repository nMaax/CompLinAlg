```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
```

```python
# Step 1: Dataset
documents = [
    "I love machine learning",
    "Machine learning is great, really great!",
    "I love coding in Python",
    "Python is great for machine learning!",
]
```

```python
# Step 2: Tokenization and Vectorization
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)
```

```python
# Display the term-document matrix
print("Term-Document Matrix:")
# Get feature names
feature_names = vectorizer.get_feature_names_out()

# Create a DataFrame to display the term-document matrix with feature names
term_document_df = pd.DataFrame(data=X.toarray(), columns=feature_names, index=documents)
display(term_document_df)
```

```python
# Step 3: Apply SVD
svd = TruncatedSVD(n_components=2)
X_svd = svd.fit_transform(X)
```

```python
# Display the SVD components
print("Reduced Data Matrix")
print(X_svd)
```

```python
# Step 2: Tokenization and Vectorization using TF-IDF
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(documents)

# Display the term-document matrix
print("Term-Document Matrix (TF-IDF):")
# Get feature names
feature_names_tfidf = tfidf_vectorizer.get_feature_names_out()

# Create a DataFrame to display the term-document matrix with feature names
term_document_df_tfidf = pd.DataFrame(data=X_tfidf.toarray(), columns=feature_names_tfidf, index=documents)
display(term_document_df_tfidf)
```

```python
# Display the top terms for each component
num_top_terms = 3
terms = vectorizer.get_feature_names_out()

for i, component in enumerate(components):
    top_terms_idx = component.argsort()[-num_top_terms:][::-1]
    top_terms = [terms[idx] for idx in top_terms_idx]
    print(f"Topic {i}: {', '.join(top_terms)}\n")
```
